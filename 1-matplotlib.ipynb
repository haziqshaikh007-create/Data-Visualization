```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Matplotlib basics

```python
x = np.linspace(0,5,11)
print(x)
y = x **3
print(y)
```

```python
plt.plot(x,y)
plt.title("hello")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

Sub-plots in matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x **2,x**2)
```

Object oriented way

```python
fig = plt.figure(figsize=(12,8),dpi=100)
axis1 = fig.add_axes([0.1,0.1,1,0.7])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.1,0.6,0.6])
axis2.plot(x,y)
```

Types of plot

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

Saving plots

```python
fig = plt.figure()
axis = fig.add_axes([0.1,0.1,0.8,0.8])
axis.plot(x,y)
plt.savefig("basicplot.png")
```

Working with Images

```python
import matplotlib.image as mpimg
```
