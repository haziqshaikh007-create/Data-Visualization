```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Heat map

```python
flights = sns.load_dataset("flights")
print(flights)
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
tipsnum= tips[["total_bill","tip","size"]]
print(tipsnum)
```

```python
corr = tipsnum.corr()
```

```python
sns.heatmap(corr, annot= True)
```

Pivot table Heat Map

```python
pvtflight = flights.pivot_table(values= "passengers", index= "month", columns= "year")
print(pvtflight)
```

```python
pvttips = tips.pivot_table(values= "tip", index= "day", columns= "size")
print(pvttips)
```

```python
plt.subplot(1,2,1)
sns.heatmap(pvtflight)
plt.subplot(1,2,2)
sns.heatmap(pvttips)
```
